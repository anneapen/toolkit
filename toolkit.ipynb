```python
import sympy as sp
```

```python
x = sp.symbols("x")
f = x**3 + 2*x
```

```python
x = sp.symbols("x")
f = sp.sin(x)
```

```python
#Derivatives
derivative = sp.diff(f,x)
print(derivative)
```

```python
#Derivative of inverse functions
x = sp.symbols("x")
# f=sp.acot(x)
# derivative = sp.diff(f,x)
derivative = sp.diff(sp.acot(x),x)
print(derivative)
```

```python
#Integrals
integral = sp.integrate(f, x)
print(integral)
```

```python
#Definite integrals
integral = sp.integrate(f,(x,0,sp.pi/2))
print(integral)
```

```python
#Quadratic equations
x=sp.symbols("x")
eqn=x**2-5*x+6
roots=sp.solve(eqn,x)
print(roots)
```

```python
x, y = sp.symbols("x y")

eq1 = sp.Eq(x + y, 10)
eq2 = sp.Eq(2*x - y, 5)

solution = sp.solve((eq1, eq2), (x, y))

print(solution)
```

```python
x, y, z = sp.symbols("x y z")

eq1 = sp.Eq(x + y + z, 6)
eq2 = sp.Eq(2*x - y + z, 3)
eq3 = sp.Eq(x + 2*y - z, 2)

solution = sp.solve((eq1, eq2, eq3), (x, y, z))

print(solution)
```
