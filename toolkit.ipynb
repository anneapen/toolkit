```python
import sympy as sp
```

```python
x = sp.symbols("x")
f = x**3 + 2*x
```

```python
x = sp.symbols("x")
f = sp.sin(x)
```

```python
derivative = sp.diff(f,x)
print(derivative)
```

```python
integral = sp.integrate(f, x)
print(integral)
```

```python
integral = sp.integrate(f,(x,0,sp.pi))
print(integral)
```
